DBSCAN

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_samples, silhouette_score
eps_array = [0.2, 0.5, 1]
min_samples_array = [5, 10, 30]

for eps in eps_array: 
    for min_sample in min_samples_array:
        cluster = DBSCAN(eps=eps, min_samples=min_sample)
        cluster_labels = cluster.fit_predict(X)

        #Skip  if only one cluster or all noise
        if len(set(cluster_labels)) <=1 or len(set(cluster_labels)) == {-1}:
            continue
        silhouette_avg = silhouette_score(X, cluster_labels)
    
        print(f"eps = {eps}, min_samples = {min_sample}")
        print(f"Number of clusters: {len(set(cluster_labels)) - (1 if -1 in cluster_labels else 0)}")
        print(f"Silhouette Score: {silhouette_avg:.3f}\n")
        
```

```python
dbscan_model = DBSCAN(eps=1, min_samples=5)
```

```python
dbscan_model.fit(X)
```

```python

print("Cluster ", len(set(dbscan_model.labels_)))
print("Score: ", silhouette_score(X, dbscan_model.labels_))
```

```python
df_work['dbscan_cluster'] = dbscan_model.labels_
```

Movie Recomendation Funciton

```python
import random

def recommend_movie(movie_name: str, n=5):
    movie_name = movie_name.lower()
    
    # match against the 'title' column (not index)
    matches = df_work[df_work['title'].str.lower().str.contains(movie_name, na=False)]
    
    if matches.empty:
        print('Movie not found in the database.')
        return []
    
    if len(matches) > 1:
        print(f"Multiple matches found, using: '{matches.iloc[0]['title']}'")
    
    movie = matches.iloc[0]
    cluster = movie['dbscan_cluster']
    
    if cluster == -1:
        print(f"'{movie['title']}' is an outlier — no similar cluster found.")
        return []
    
    cluster_movies = df_work[
        (df_work['dbscan_cluster'] == cluster) & (df_work['title'] != movie['title'])
    ]
    
    if cluster_movies.empty:
        print(f"No other movies found in the same cluster as '{movie['title']}'.")
        return []
    
    n = min(n, len(cluster_movies))
    recommended_titles = random.sample(list(cluster_movies['title']), n)
    
    print('--- We can recommend you these movies ---')
    for m in recommended_titles:
        print(m)
    
    return recommended_titles
```

```python
input_movie = input("Enter the movie")
recommend_movie(input_movie)
```
